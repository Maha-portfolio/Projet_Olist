```python
#Vérifier l'unicité
cles_candidates = {
    "orders": ["order_id"],
    "customers": ["customer_id"],
    "product": ["product_id"],
    "sellers": ["seller_id"],
    "product_category": ["product_category_name"],
    "order_reviews": ["review_id"],
    "order_payment": ["order_id", "payment_sequential"],
    "orders_item": ["order_id", "order_item_id"]
}

#Une fonction de vérification

def verifier_unicite(nom, df, cles):

    lignes = len(df)

    valeurs_manquantes = df[cles].isna().any(axis=1).sum()

    doublons = df.duplicated(subset=cles).sum()

    combinaisons_distinctes = df[cles].drop_duplicates().shape[0]

    print(f"\n--- {nom} ---")
    print("Nombre de lignes :", lignes)
    print("Clé(s) testée(s) :", cles)
    print("Lignes avec clé(s) manquante(s) :", valeurs_manquantes)
    print("Combinaisons dupliquées :", doublons)
    print("Combinaisons distinctes :", combinaisons_distinctes)

    if valeurs_manquantes == 0 and doublons == 0:
        print("Résultat : clé candidate valide pour l'unicité.")
    else:
        print("Résultat : investigation nécessaire.")

```

```python
for nom, cles in cles_candidates.items():
    verifier_unicite(nom, tables[nom], cles)
```
